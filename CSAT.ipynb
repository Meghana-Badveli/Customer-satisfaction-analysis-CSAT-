```python
for col in df.select_dtypes(include='object'):
  df[col]=df[col].str.strip()
```

```python
df['call_timestamp'] = pd.to_datetime(
    df['call_timestamp'], errors='coerce'
)
```

```python
df['csat_score'] = pd.to_numeric(
    df['csat_score'], errors='coerce'
)
```
